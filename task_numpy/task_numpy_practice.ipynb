* __Задача 1__: Подсчитать произведение ненулевых элементов на диагонали прямоугольной матрицы.

```python
reload_solution_modules()
time_non_vectorized = []
time_vectorized = []
data_size = [10, 30, 70, 120, 1000, 5000, 10000]
for n in data_size:
    x = np.random.randint(0, 10, [n, n])


    time_start = time.time()
    functions_vectorized.prod_non_zero_diag(x)
    time_finish = time.time()
    time_vectorized.append(time_finish-time_start)

    time_start = time.time()
    functions.prod_non_zero_diag(x.tolist())
    time_finish = time.time()
    time_non_vectorized.append(time_finish-time_start)



figure, axis = plt.subplots(1, 1, figsize=(13, 5))

axis.plot(data_size, time_non_vectorized, label="Невекторизованная")
axis.plot(data_size, time_vectorized, label="Векторизованная")
axis.set_title("График векторизованной и не векторизованной функции")
axis.set_xlabel("Длина массива")
axis.set_ylabel("Время (с)")
axis.grid(True, alpha=0.3)
axis.legend()

figure.tight_layout()
plt.show()
```
